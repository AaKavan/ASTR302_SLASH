```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
### names=True uses the first line as column names.
data26 = np.genfromtxt("data/text_files/SN2026aaiv.txt", names=True, dtype=None, encoding=None)

# Pull out each column
days26 = data26["mjd"]
mags26 = data26["mag"]
filters26 = data26["filter"]


# Keep only the points for each filter
B_days26, B_mags26 = days26[filters26 == "B"], mags26[filters26 == "B"]
V_days26, V_mags26 = days26[filters26 == "V"], mags26[filters26 == "V"]
g_days26, g_mags26 = days26[filters26 == "gp"], mags26[filters26 == "gp"]
r_days26, r_mags26 = days26[filters26 == "rp"], mags26[filters26 == "rp"]
i_days26, i_mags26 = days26[filters26 == "ip"], mags26[filters26 == "ip"]

### Plot each filter as dots
fig, ax = plt.subplots()
ax.plot(B_days26, B_mags26, "o", label="B")
ax.plot(V_days26, V_mags26, "o", label="V")
ax.plot(g_days26, g_mags26, "o", label="g")
ax.plot(r_days26, r_mags26, "o", label="r")
ax.plot(i_days26, i_mags26, "o", label="i")

### invert y axis
ax.invert_yaxis()
ax.set_xlabel("Days (MJD)")
ax.set_ylabel("Magnitude")
ax.set_title("SN2026aaiv")
ax.legend()
plt.show()
```
