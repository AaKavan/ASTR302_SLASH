```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
### names=True uses the first line as column names.
data = np.genfromtxt("data/SN2026aaiv.txt", names=True, dtype=None, encoding=None)

# Pull out each column
days = data["mjd"]
mags = data["mag"]
filters = data["filter"]


# Keep only the points for each filter
B_days, B_mags = days[filters == "B"], mags[filters == "B"]
V_days, V_mags = days[filters == "V"], mags[filters == "V"]
g_days, g_mags = days[filters == "gp"], mags[filters == "gp"]
r_days, r_mags = days[filters == "rp"], mags[filters == "rp"]
i_days, i_mags = days[filters == "ip"], mags[filters == "ip"]

### Plot each filter as dots
fig, ax = plt.subplots()
ax.plot(B_days, B_mags, "o", label="B")
ax.plot(V_days, V_mags, "o", label="V")
ax.plot(g_days, g_mags, "o", label="g")
ax.plot(r_days, r_mags, "o", label="r")
ax.plot(i_days, i_mags, "o", label="i")

### invert y axis
plt.gca().invert_yaxis()  
plt.xlabel("Days (MJD)")
plt.ylabel("Magnitude")
plt.legend()
plt.show()

```
