```python
import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits
import glob, os
import pandas as pd
from astropy.time import Time
```

```python
### All the w files from Nov 3
files = sorted(glob.glob("data/slash/03nov26/w*.fits"))

### Print the important header info for each one
for f in files:
    h = fits.getheader(f)
    print(f, h["OBJECT"], h["FILTER"], h["EXPTIME"])
```

```python
### We have to separate the files by target, since two groups shared one night
sn2026aaiv_files = []
for f in files:
    obj = fits.getheader(f)["OBJECT"].lower()
    if "sqf" in obj:
        sn2026aaiv_files.append(f)

print(len(sn2026aaiv_files), "sqf files")
sn2026aaiv_files
```
